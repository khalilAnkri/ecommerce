## Import libraries

```python
import numpy as np
import pandas as pd
from apyori import apriori
```

## Load dataset

```python
store_data = pd.read_csv('../data/store_data.csv', header=None)
print(store_data.head())
```

`header=None` tells pandas that the CSV file does not contain a header row with column names.

This ensures the first line of actual data is **not mistaken** for column names.

## Filter out 'nan' values

```python
records = []
for i in range(0, 7501): #OR for i in range(len(store_data)):
    records.append([str(item) for item in store_data.values[i] if str(item) != 'nan'])
```

```python
records[:2]
```

## Define and run the algorithm

```python
association_rules = apriori(records, min_support=0.0045, min_confidence=0.2, min_lift=3, min_length=2)
association_results = list(association_rules)
association_results
```

- Support = freq(A,B)/N


- Confidence = freq(A,B)/freq(A)

-> How often A, B occur together, given number of times A occur


- Lift = support/supp(A).supp(B)

-> independece occurance of A,B

## Sorting results based on their confidence values

```python
listRules = sorted(
    [list(rule.ordered_statistics) for rule in association_results],
    key=lambda x: x[0].confidence,
    reverse=True
)
```

```python
listRules[0]
```

`reverse=True` → from highest to lowest

```python
students = [ ["Alice", 85],    # Index 0
             ["Bob", 92],      # Index 1
            ["Charlie", 78]   # Index 2
             ]
sorted_students = sorted(students, key=lambda x: x[1], reverse=True)
sorted_students
```

## Enumerate

```python
for idx, rl in enumerate(listRules):
  print(f'\nRule{idx+1}')
  for statistic in rl:
    print(f'Item Base: {statistic.items_base}')
    print(f'Item Add: {statistic.items_add}')
    print(f'Confidence: {statistic.confidence}')
    print(f'Lift: {statistic.lift}')
    print('---------------------------------------------')
```
